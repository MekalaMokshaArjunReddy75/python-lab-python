```python
def palin(n):
  original=n
  rev=0
  while(n>0):
    r=n%10
    rev=rev*10+r
    n=n//10
  if rev==original:
    print("palindrome")
  else:
    print("not a palindrome")
n=int(input("enter a number"))
palin(n)
```

```python
def rev(n):
  rev=0
  while(n>0):
    r=n%10
    rev=rev*10+r
    n=n//10
  return rev
n=int(input("enter a number"))
rev(n)
```

```python
def rev_str(s):
  rev=s[::-1]
  return rev
s=input()
rev_str(s)
```

```python
def revers_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
s=input()
revers_str(s)
```

```python
#EXAMPLE ON DEFAULT ARGUMENTS
def print_msg(name,branch="aids"):
  print("name is",name,"branch is ",branch)
print_msg("surya")
```

```python
#EXAMPLE ON KEYWOARD ARGUMENT
def print_msg(name,branch):
  print("name is",name,"branch is ",branch)
print_msg(branch="AIDS",name="SURYA")
```

```python
def add(*num):
  print(num)
  print(sum(num))
add(1,2,34,5)
```

```python
#FUNCTIONS THAT RETURNS MULTIPLE VALUES
def multiple_values(*nums):
  s=sum(nums)
  c=len(nums)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4)
print(f"sum is {res_sum},count is {res_count},the average is {res_avg}")
```
